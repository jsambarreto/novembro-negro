```python
%%writefile app.py
import streamlit as st
import pandas as pd

st.set_page_config(page_title="Painel Novembro Negro", page_icon="📊", layout="wide")

@st.cache_data
def carregar_dados():
    dados = {
        "Ano": [2019, 2019, 2020, 2020, 2021, 2021, 2022, 2022],
        "Cor_Raca": ["Branca", "Preta/Parda", "Branca", "Preta/Parda", "Branca", "Preta/Parda", "Branca", "Preta/Parda"],
        "Rendimento_Medio": [2900, 1600, 2850, 1550, 2800, 1500, 3100, 1750],
        "Regiao": ["Nordeste"] * 8
    }
    return pd.DataFrame(dados)

df = carregar_dados()

st.title("Axé no Sertão: Visibilidade e Dados ✊🏾")
st.markdown("Análise da disparidade socioeconômica no Brasil utilizando dados públicos.")

st.sidebar.header("Filtros de Análise")
ano_selecionado = st.sidebar.selectbox("Selecione o Ano:", df["Ano"].unique())
df_filtrado = df[df["Ano"] == ano_selecionado]

st.subheader(f"Rendimento Médio no Nordeste em {ano_selecionado}")
st.dataframe(df_filtrado, use_container_width=True)

st.subheader("Evolução Histórica do Rendimento")
df_grafico = df.pivot(index="Ano", columns="Cor_Raca", values="Rendimento_Medio")
st.line_chart(df_grafico)
```
